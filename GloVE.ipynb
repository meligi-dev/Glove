```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import matplotlib.pyplot as plt
```

```python
corpus = ["apple banana fruit", "banana apple fruit", "banana fruit apple",
                 "dog cat animal", "cat animal dog", "cat dog animal","grape apple apple","fish dog dog"]
```

```python
corpus = [sent.split(" ") for sent in corpus]
corpus
```

```python
flatten = lambda l: [item for sublist in l for item in sublist]
vocab = list(set(flatten(corpus)))
vocab
```

```python
word2indx = {v : indx for indx , v in enumerate(vocab)}
indx2word = {indx : v for indx , v in enumerate(vocab)}
```

```python
vocab.append("<UNK>")
word2indx["<UNK>"] = len(vocab) - 1
indx2word[len(vocab) - 1] = "<UNK>"
```

```python
def Random_batch(batch_size,corpus):
    skipgram = []
    for doc in corpus:
        for i in range(1,len(doc)-1):
            center = word2indx[doc[i]]
            outside = (word2indx[doc[i-1]],word2indx[doc[i+1]])
            for each_outside in outside:
                skipgram.append([center,each_outside])
    random_index = np.random.choice(range(len(skipgram)),batch_size,replace = False)
    inputs,label = [],[]
    for indx in random_index:
        inputs.append([skipgram[indx][0]])
        label.append([skipgram[indx][1]])
    return np.array(inputs),np.array(label)
```

```python
from collections import Counter
skipgram = []

for sent in corpus:
    for i in range(1,len(sent)-1):
        center = sent[i]
        outside = [sent[i-1],sent[i+1]]
        for w in outside:
            skipgram.append((center,w))
counter_skipgram = Counter(skipgram)
counter_skipgram
```

```python
def weigthing(w_i,w_j,counter_skipgram):
    try:
        xi_j = counter_skipgram((w_i,w_j))
    except:
        xi_j = 1
    x_max = 100
    alpha = 0.75
    if xi_j < x_max:
        result = (xi_j / x_max) ** alpha
    else:
        result = 1
    return result
```
